build a 4D array (batch, height, width, channels), reshape it, transpose axes, flatten it back.

```python
import numpy as np
```

```python
arr = np.arange(8*4*8*1).reshape(8,4,8,1)
print(arr.shape, arr.ndim, arr.dtype, arr.size)
```

```python
arr = arr.transpose(3, 1, 2, 0)
print(arr.shape, arr.ndim, arr.dtype, arr.size)
```

```python
arr.flags['C_CONTIGUOUS'] 
```

```python
arr = arr.ravel()
print(arr.shape, arr.ndim, arr.dtype, arr.size)
```
