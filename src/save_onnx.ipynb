```python
import gymnasium as gym
import torch

from src.inference.cart_pole import load_agent_a2c

```

```python
env = gym.make("CartPole-v1", render_mode=None)
state, _ = env.reset()

n_observations = len(state)
n_actions = env.action_space.n

agent = load_agent_a2c(
    "./data/ppo_lr0.00025_ne6000_k4_r1_20260820_214809.safetensors",
    n_observations=n_observations,
    n_actions=n_actions,
)
```

```python
dummy_input = torch.randn(1, 4)  # batch=1, 4 obs dims (cart pos, vel, pole angle, angular vel)
torch.onnx.export(
    agent,
    dummy_input,
    "./data/cartpole_ppo_lr0.00025_ne6000_k4_r1_20260820_214809.onnx",
    input_names=["obs"],
    output_names=["action_probs"],
    dynamic_axes={"obs": {0: "batch"}, "action_probs": {0: "batch"}},
    external_data=False,
)
```
